## What is an API?
APIs (Application Programming Interfaces) enable software applications to communicate with each other by sending and receiving data. For example, phone apps use APIs to retrieve live information like traffic conditions, weather reports, or sports scores. Often, an API request requires an 'API Key,' a unique identifier that grants access to the API. API Keys should always be kept private and never shared, as they control access to the API.

The United States Census Bureau allows anyone to request an API Key to allow programmatic requests for census data. To request your own key (necessary for the following exhibits), visit https://api.census.gov/data/key_signup.html

```python
import requests
import pandas as pd

# Replace with your actual Census API key
api_key = 'yourCensusAPIKey'
```

```python

# The Census Bureau API URL for 2020 Decennial Census data
url = 'https://api.census.gov/data/2020/dec/pl'
```

```python
# Parameters for getting the population data for each county in Ohio (state FIPS code is 39)
params = {
    'get': 'NAME,P1_001N',  # Request the county name (NAME) and total population (P1_001N)
    'for': 'county:*',  # Get data for all counties in Ohio
    'in': 'state:39',  # Ohio's state FIPS code is 39
    'key': api_key  # Your Census API key
}

# Parse the response as JSON
data = requests.get(url, params=params).json()

# Create an empty list to store county data
counties_data = []

# Loop through the counties and extract their names and populations
for row in data[1:]:  # Skip the first row, which contains column names
    county_name = row[0]
    population = row[1]
    counties_data.append([county_name, population])  # Store the data as a list of lists

# Create a Pandas DataFrame from the list of data
df = pd.DataFrame(counties_data, columns=['County', 'Population'])

# Display the DataFrame
df
```
