# Object-oriented Programming

```python
# Everything is an object.

class Animal:
    def __init__(self, species, sound):
        self.species = species
        self.sound = sound
        self.hunger = 50

    def eat(self):
        if self.hunger > 0:
            self.hunger -= 50
            self.hunger = max(self.hunger, 0)
            print(f"{self.species} is eating. Hunger level: {self.hunger}")
        else:
            print(f"{self.species} is not hungry. Hunger level: {self.hunger}")

    def play(self):
        if self.hunger > 80:
            print(f"{self.species} is too hungry to play. Hunger level: {self.hunger}")
        else:
            self.hunger += 20
            self.hunger = min(self.hunger, 100)
            print(f"{self.species} is playing. Hunger level: {self.hunger}")

    def make_sound(self):
        print(f"{self.species} says {self.sound}.")
```

```python
dog = Animal("Dog", "Woof")
dog.make_sound()
```

```python
dog.eat()
```

```python
dog.play()
```
